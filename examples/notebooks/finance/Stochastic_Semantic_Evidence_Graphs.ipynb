# BeliefLens stochastic semantic evidence graphs

This notebook shows how BeliefLens treats a retrieval-augmented workflow as an auditable sequence of channels rather than one opaque answer. It uses frozen public outputs, so the first part requires no API key and makes no model-provider calls.

The graph is: **source evidence -> retrieved evidence -> prompt and model response -> semantic state -> qualified decision input**.

A graph certificate is conditional on the declared benchmark, ontology, model/service version and error tolerances. It is not a blanket guarantee that an AI system is correct.

```python
import os
import httpx
import pandas as pd

BELIEFLENS_URL = os.environ.get('BELIEFLENS_URL', 'https://demo.belieflens.org')
client = httpx.Client(base_url=BELIEFLENS_URL, timeout=60)

# Immutable public example: cached and provider-call free.
example = client.get('/v1/examples/SPY-trading').json()
pd.Series({
    'endpoint': example.get('endpoint'),
    'demo mode': example.get('demo_mode'),
    'provider API calls': example.get('provider_api_calls'),
    'reproduction passed': example.get('reproduction_passed'),
}, name='frozen demonstration')
```

## The semantic evidence graph

A workflow is governed through five checks:

1. **Evidence and retrieval:** was the retrieved material sufficient, relevant and non-contradictory for the declared task?
2. **Prompt channel:** do information-equivalent prompt presentations preserve the semantic measurement?
3. **Response channel:** what probability mass does the model assign to complete candidate continuations?
4. **Semantic state:** how do those continuation probabilities map to the declared business states?
5. **Decision qualification:** do the held-out calibration, coverage and stability tests support use, review or abstention?

The public attribution route below illustrates the diagnostic layer. It reports uncertainty contributions from the frozen example without calling a provider.

```python
attribution = client.get('/v1/examples/SPY-trading/attributions').json()
pd.Series({
    'endpoint': attribution.get('endpoint'),
    'provider API calls': attribution.get('provider_api_calls'),
    'analysis type': attribution.get('analysis_type'),
    'top-k requested': attribution.get('top_k'),
}, name='diagnostic result')
```

## Certificate interpretation

A pass means: *within this frozen benchmark and stated tolerances, the workflow met its declared retrieval, stability, calibration and coverage gates.*

An abstention or fail is useful information: it tells the operator not to promote the model output as a qualified decision-state probability until the responsible channel is investigated.
